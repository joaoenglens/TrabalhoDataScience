```python
import pandas as pd
import numpy as np
```

```python
renda = pd.read_csv("C:/Users/tomas/OneDrive/Desktop/DStrabalho/data/processed/Nota_Renda.csv",
                    sep=";")
```

```python
print(renda.columns)
```

```python
print(renda["RENDA"].unique())
```

```python
mapa_2009_2010 = {
 "A" : "1",
 "B" : "1.5",
 "C" : "3.5",
 "D" : "7.5",
 "E" : "20",
 "F" : "40",
 "G" : "55",
 "H" : "0"
}

mapa_2011 ={
    "A" : "0",
    "B" : "1",
    "C" : "1.25",
    "D" : "1.75",
    "E" : "3.5",
    "F" : "6",
    "G" : "8.5",
    "H" : "11",
    "I" : "13.5",
    "J" : "22.5",
    "K" : "35"
}

mapa_2012_2020 = {
        "A" : "0",
        "B" : "1",
        "C" : "1.25",
        "D" : "1.75",
        "E" : "2.25",
        "F" : "2.75",
        "G" : "3.5",
        "H" : "4.5",
        "I" : "5.5",
        "J" : "6.5",
        "K" : "7.5",
        "L" : "8.5",
        "M" : "9.5",
        "N" : "11",
        "O" : "13.5",
        "P" : "17.5",
        "Q" : "25"
}
```

```python
renda["ANO"] = pd.to_numeric(renda["ANO"], errors="coerce")

nova_renda = (
    renda["RENDA"]
    .astype("string")
    .str.strip()
    .str.upper()
)
```

```python
periodo_1 = renda["ANO"].between(2009, 2010)
periodo_2 = renda["ANO"].eq(2011)
periodo_3 = renda["ANO"].between(2012, 2020)
```

```python
renda["salario"] = pd.Series(
    pd.NA,
    index=renda.index,
    dtype="Float64"
)

renda.loc[periodo_1, "salario"] = (
    nova_renda.loc[periodo_1].map(mapa_2009_2010)
)

renda.loc[periodo_2, "salario"] = (
    nova_renda.loc[periodo_2].map(mapa_2011)
)

renda.loc[periodo_3, "salario"] = (
    nova_renda.loc[periodo_3].map(mapa_2012_2020)
)
```

```python
print(renda[["ANO","RENDA","salario"]])
```

```python
nao_convertidos = renda.loc[
    renda["RENDA"].isna() & renda["salario"].notna(),
    ["ANO", "RENDA"]
].drop_duplicates()

print(nao_convertidos)
```

```python
renda.to_csv( "NOTA_RENDA_N.csv",
    encoding="utf-8",
        sep=";",
        index=False
)
```
