Agora vamos efetuar teste de hipotese

Vamos verificar se o aumento de renda realmente aumenta a nota 

Para isso selecionamos um p-valor de 0.05

H0 define que o aumento de renda não aumenta as notas dos estudandes 

H1 o aumento de renda aumenta as notas 


```python
x = df_periodo["salario"]
y = df_periodo["MEDIA_NOTA"]

plt.figure(figsize=(16,12))
plt.scatter(
    
    x,
    y,
    color = "royalblue",
    alpha = 0.5,
    label = "Estudante",
    s=20
)

plt.title("Notas X Salarios",fontsize=15)
plt.xlabel("Salários")
plt.ylabel("Notas")
plt.legend()
plt.show()
```

```python
import numpy as np
import statsmodels.api as sm
from scipy.stats import t

dados = df_periodo[["salario", "MEDIA_NOTA"]].copy()
dados = dados.apply(pd.to_numeric, errors="coerce")
dados = dados.replace([np.inf, -np.inf], np.nan).dropna()

if len(dados) < 3 or dados["salario"].nunique() < 2:
    raise ValueError("É necessário ter pelo menos 3 registros e 2 salários distintos.")
```

```python
resumo = (
    dados.groupby("salario", as_index=False)
    .agg(
        media_nota=("MEDIA_NOTA", "mean"),
        desvio=("MEDIA_NOTA", "std"),
        quantidade=("MEDIA_NOTA", "size")
    )
)

resumo["erro_95"] = (
    t.ppf(0.975, resumo["quantidade"] - 1)
    * resumo["desvio"]
    / np.sqrt(resumo["quantidade"])
)

resumo
```

```python
X = sm.add_constant(dados[["salario"]], has_constant="add")
modelo = sm.OLS(dados["MEDIA_NOTA"], X).fit(cov_type="HC3")

# H1: coeficiente do salário > 0
beta = modelo.params["salario"]
p_bilateral = modelo.pvalues["salario"]
p_unilateral = p_bilateral / 2 if beta >= 0 else 1 - p_bilateral / 2

# Evita exibir p-valor muito pequeno como "0.0000"
texto_p = (
    "p-valor unilateral < 1e-300"
    if p_unilateral < 1e-300
    else f"p-valor unilateral = {p_unilateral:.3g}"
)

print(modelo.summary())
```

```python


x_linha = np.linspace(dados["salario"].min(), dados["salario"].max(), 200)

X_linha = sm.add_constant(
    pd.DataFrame({"salario": x_linha}),
    has_constant="add"
)

previsoes = modelo.get_prediction(X_linha).summary_frame(alpha=0.05)

 
fig, ax = plt.subplots(figsize=(10, 6))

com_ic = resumo[resumo["quantidade"] > 1]

ax.errorbar(
    com_ic["salario"],
    com_ic["media_nota"],
    yerr=com_ic["erro_95"],
    fmt="none",
    ecolor="royalblue",
    capsize=4,
    alpha=0.7,
    label="IC de 95% das médias"
)

# Pontos maiores indicam faixas com mais estudantes
tamanhos = 40 + 200 * np.sqrt(
    resumo["quantidade"] / resumo["quantidade"].max()
)

ax.scatter(
    resumo["salario"],
    resumo["media_nota"],
    s=tamanhos,
    color="royalblue",
    edgecolor="white",
    label="Média por faixa salarial",
    zorder=3
)

ax.plot(
    x_linha,
    previsoes["mean"].to_numpy(),
    color="tomato",
    linewidth=2,
    label="Regressão linear"
)

ax.fill_between(
    x_linha,
    previsoes["mean_ci_lower"].to_numpy(),
    previsoes["mean_ci_upper"].to_numpy(),
    color="tomato",
    alpha=0.15,
    label="IC de 95% da regressão"
)

ax.text(
    0.03, 0.97,
    f"Inclinação = {beta:.3f}\n{texto_p}",
    transform=ax.transAxes,
    va="top",
    bbox=dict(facecolor="white", edgecolor="gray", alpha=0.9)
)

ax.set_xlabel("Renda representativa da faixa (salários mínimos)")
ax.set_ylabel("Média das notas")
ax.set_title("Notas por faixa de renda")
ax.legend(loc="lower right")
ax.grid(alpha=0.2)
fig.tight_layout()
plt.show()
```
