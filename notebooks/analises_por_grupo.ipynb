Agora vamos fazer um mapinha de media por estado

```python
media_uf = df_periodo.groupby('UF', observed=True)['MEDIA_NOTA'].mean().reset_index()

print(df_periodo['UF'].unique())
```

```python
import plotly.express as px
import json
import urllib.request

url = 'https://raw.githubusercontent.com/codeforamerica/click_that_hood/master/public/data/brazil-states.geojson'
with urllib.request.urlopen(url) as response:
    geojson_brasil = json.load(response)

fig = px.choropleth(
    media_uf,
    geojson=geojson_brasil,
    locations='UF',
    featureidkey='properties.sigla',
    color='MEDIA_NOTA',
    color_continuous_scale='RdYlGn',
    scope='south america',
    labels={'MEDIA_NOTA': 'Nota média'}
)
fig.update_geos(fitbounds='locations', visible=False)
fig.update_layout(title='Nota média do ENEM por estado (2015-2020)')
fig.write_html("mapinha.html")
```
